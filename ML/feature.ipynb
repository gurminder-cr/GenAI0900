```python
data= pd.DataFrame({'Result':['good','bad','good','bad','excellent','good','good','excellent']})
data
```

```python
# data.Result.map({'good':1,'bad':2,'excellent':3})
```

```python
# label encoder 
from sklearn.preprocessing import LabelEncoder
encoder= LabelEncoder()
```

```python
# fit, transform, fit_transform 
encoder.fit(data.Result)
```

```python
encoder.transform(data.Result)
```

```python
data['Label_result']=encoder.transform(data.Result)
```
