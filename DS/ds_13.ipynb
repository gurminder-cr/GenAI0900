Exploratory Data Analysis (EDA) is an important step in data science and data analytics as it visualizes data to understand its main features, find patterns and discover how different parts of the data are connected.

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns 
import plotly.express as px 
```

```python
df= pd.read_csv('zomato_dataset.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.dropna(inplace=True)
```

```python
df.shape
```

```python
df['Restaurant Name'].value_counts()
```

```python
df.City.value_counts()
```

```python
city_Labels= df.City.value_counts().index
city_Labels[:5]
```

```python
city_values= df.City.value_counts().values 
city_values[:5]
```

```python
plt.pie(x=city_values[:5],labels=city_Labels[:5], autopct="%1.0f%%")
plt.show()
```

```python
px.pie(names=city_Labels[:10], values=city_values[:10])
```

```python
df.columns
```

```python
df['Cuisine '].value_counts()
```

```python
df.groupby(by='Cuisine ').get_group('Tibetan')
```

```python
df.groupby(by='Cuisine ')['Votes'].sum().sort_values(ascending=False)
```

```python
cuisine_values= df.groupby(by='Cuisine ')['Votes'].sum().sort_values(ascending=False).values
cuisine_Labels=df.groupby(by='Cuisine ')['Votes'].sum().sort_values(ascending=False).index
```

```python
px.bar(x=cuisine_Labels[:5], y=cuisine_values[:5], color=cuisine_Labels[:5])
```
