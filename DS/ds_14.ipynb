```python
import pandas as pd
import numpy as np
```

```python
# map, applymap, getdummies

df=pd.DataFrame({'Name':['Sourav','Priyanshu','Simran','Anshu','Ansh','Anurag','Priyanka'],'class':['btech','bca','mca','btech','mcom','mba','mca'],'Rollno':[34,23,11,10,11,67,89]})
df
```

```python
df.loc[:,'class'].str.capitalize()
```

```python
df.loc[:,'Rollno']= df.loc[:,'Rollno']+2
```

```python
df
```

```python
def funcCap(text):
    return text.capitalize()
```

```python
df.loc[:,'class']= df.loc[:,'class'].apply(funcCap)  # series
```

```python
df
```

```python
df.map(lambda x: str(x)+' hi')
```
